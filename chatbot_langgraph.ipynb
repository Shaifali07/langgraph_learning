```python
from langgraph.graph import StateGraph, START,END
from typing import TypedDict, Annotated
from pydantic import BaseModel,fields
from langchain_groq import ChatGroq
```

```python
from google.colab import userdata
import os

groq_api_key = userdata.get('GROQ_API_KEY')
os.environ["GROQ_API_KEY"] = groq_api_key
```

```python
llm=ChatGroq(
    model_name="llama-3.1-8b-instant",
    temperature=0)
```

```python
from langgraph.graph.message import add_messages
class chat_bot(TypedDict):
  messages: Annotated[list,add_messages]
```

```python
def get_message(state:chat_bot):
  response=llm.invoke(state["messages"])
  return {"messages":response}


```

```python
graph=StateGraph(chat_bot)
graph.add_node("get_message",get_message)
graph.add_edge(START,"get_message")
graph.add_edge("get_message",END)

initial_state={
    "messages":"what is capital of India?"
}
workflow=graph.compile()
final_state=workflow.invoke(initial_state)
print(final_state)
```
